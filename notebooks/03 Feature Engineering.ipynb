# Feature Engineering

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
customers = pd.read_csv("../data/cleaned/customers_cleaned.csv")
orders = pd.read_csv("../data/cleaned/orders_cleaned.csv")
orderitems = pd.read_csv("../data/cleaned/orderitems_cleaned.csv")
products = pd.read_csv("../data/cleaned/products_cleaned.csv")
payments = pd.read_csv("../data/cleaned/payments_cleaned.csv")
reviews = pd.read_csv("../data/cleaned/reviews_cleaned.csv")
sellers = pd.read_csv("../data/cleaned/sellers_cleaned.csv")
categories = pd.read_csv("../data/cleaned/categories_cleaned.csv")
geolocation = pd.read_csv("../data/cleaned/geolocation_cleaned.csv")
```

## Orders Dataset

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

orders[date_columns] = orders[date_columns].apply(pd.to_datetime)
```

```python
orders["purchase_year"] = orders["order_purchase_timestamp"].dt.year
```

```python
orders["purchase_month"] = orders["order_purchase_timestamp"].dt.month
```

```python
orders["purchase_month_name"] = (
    orders["order_purchase_timestamp"]
    .dt.month_name()
)
```

```python
orders["purchase_day"] = orders["order_purchase_timestamp"].dt.day
```

```python
orders["purchase_weekday"] = (
    orders["order_purchase_timestamp"]
    .dt.day_name()
)
```

```python
orders["purchase_hour"] = orders["order_purchase_timestamp"].dt.hour
```

```python
orders.head()
```

The Orders dataset was enriched with multiple time-based features including purchase year, month, month name, day, weekday, and hour. These attributes will support seasonal trend analysis, customer purchasing behavior, and sales pattern exploration.

## Reviews Dataset

```python
date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

reviews[date_columns] = reviews[date_columns].apply(pd.to_datetime)
```

```python
reviews["review_response_days"] = (
    reviews["review_answer_timestamp"] -
    reviews["review_creation_date"]
).dt.days
```

```python
reviews.head()
```

The review response time feature was created to measure the number of days taken to respond to customer reviews. This feature can be used to evaluate customer service efficiency.

## Payments Dataset

```python
payments["payment_category"] = payments["payment_type"].replace({
    "credit_card": "Card",
    "debit_card": "Card",
    "voucher": "Voucher",
    "boleto": "Bank Transfer"
})
```

```python
payments.head()
```

Payment methods were grouped into broader business categories to simplify reporting and payment trend analysis.

## Products Dataset

```python
products["product_volume_cm3"] = (
    products["product_length_cm"] *
    products["product_height_cm"] *
    products["product_width_cm"]
)
```

```python
products.head()
```

A new product volume feature was created using the product dimensions. This feature can be used to analyze shipping costs, storage requirements, and logistics performance.

## Customers Dataset

```python
customers["customer_location"] = (
    customers["customer_city"] +
    ", " +
    customers["customer_state"]
)
```

```python
customers.head()
```

A customer location feature was created by combining city and state information. This simplifies regional analysis and reporting.

## Save Engineered Datasets

```python
Path("../data/engineered").mkdir(parents=True, exist_ok=True)
```

```python
customers.to_csv("../data/engineered/customers_engineered.csv", index=False)
orders.to_csv("../data/engineered/orders_engineered.csv", index=False)
orderitems.to_csv("../data/engineered/orderitems_engineered.csv", index=False)
products.to_csv("../data/engineered/products_engineered.csv", index=False)
payments.to_csv("../data/engineered/payments_engineered.csv", index=False)
reviews.to_csv("../data/engineered/reviews_engineered.csv", index=False)
sellers.to_csv("../data/engineered/sellers_engineered.csv", index=False)
categories.to_csv("../data/engineered/categories_engineered.csv", index=False)
geolocation.to_csv("../data/engineered/geolocation_engineered.csv", index=False)
```
