# Data Cleaning

```python
import pandas as pd
import numpy as np
```

```python
customers = pd.read_csv("../data/raw/olist dataset/customers.csv")
orders = pd.read_csv("../data/raw/olist dataset/orders.csv")
orderitems = pd.read_csv("../data/raw/olist dataset/orderitems.csv")
products = pd.read_csv("../data/raw/olist dataset/products.csv")
payments = pd.read_csv("../data/raw/olist dataset/payments.csv")
reviews = pd.read_csv("../data/raw/olist dataset/reviews.csv")
sellers = pd.read_csv("../data/raw/olist dataset/sellers.csv")
categories = pd.read_csv("../data/raw/olist dataset/categories.csv")
geolocation = pd.read_csv("../data/raw/olist dataset/geolocation.csv")
```

## Customers Dataset

```python
customers.dtypes
```

The customers dataset is already clean. No missing values, duplicate records, or data type conversions are required.

## Orders Dataset

```python
orders.dtypes
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

orders[date_columns] = orders[date_columns].apply(pd.to_datetime)
```

```python
orders.dtypes
```

All order date columns were successfully converted to datetime format. Missing delivery and approval dates were retained because they represent legitimate business scenarios such as cancelled or undelivered orders.

## Order Items Dataset

```python
orderitems.dtypes
```

```python
orderitems[["price", "freight_value"]].describe()
```

```python
(orderitems[["price", "freight_value"]] < 0).sum()
```

The Order Items dataset contains valid numeric values. No negative prices or freight charges were found, and no additional cleaning is required.

## Products Dataset

```python
products.dtypes
```

```python
products.isnull().sum()
```

```python
products["product_category_name"] = products["product_category_name"].fillna("Unknown")
```

```python
numeric_columns = [
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

products[numeric_columns] = products[numeric_columns].fillna(
    products[numeric_columns].median()
)
```

```python
products.isnull().sum()
```

Missing product category names were replaced with "Unknown". Missing numerical values were imputed using the median to preserve the overall distribution while minimizing the impact of outliers.

## Payments Dataset

```python
payments.dtypes
```

```python
payments["payment_value"].describe()
```

```python
(payments["payment_value"] < 0).sum()
```

```python
(payments["payment_installments"] < 0).sum()
```

The Payments dataset is clean. Payment values and installment counts are valid, with no negative values detected. No additional data cleaning is required.

## Reviews Dataset

```python
reviews.dtypes
```

```python
date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

reviews[date_columns] = reviews[date_columns].apply(pd.to_datetime)
```

```python
reviews.dtypes
```

Review date columns were converted to datetime format. Missing values in review titles and review messages were retained because many customers submitted only ratings without providing written feedback. These missing values represent valid business scenarios and were not treated as data quality issues.

## Sellers Dataset

```python
sellers.dtypes
```

The Sellers dataset is already clean. No missing values, duplicate records, or data type conversions are required.

## Categories Dataset

```python
categories.dtypes
```

The Categories dataset is complete and requires no preprocessing. The mapping between Portuguese and English category names is ready for downstream analysis.

## Geolocation Dataset

```python
geolocation.dtypes
```

```python
geolocation = geolocation.drop_duplicates()
```

```python
geolocation.duplicated().sum()
```

Duplicate geolocation records were removed to eliminate redundant location entries. No missing values or invalid coordinate values were detected.

## Save Cleaned Datasets

```python
from pathlib import Path

Path("../data/cleaned").mkdir(parents=True, exist_ok=True)
```

```python
customers.to_csv("../data/cleaned/customers_cleaned.csv", index=False)
orders.to_csv("../data/cleaned/orders_cleaned.csv", index=False)
orderitems.to_csv("../data/cleaned/orderitems_cleaned.csv", index=False)
products.to_csv("../data/cleaned/products_cleaned.csv", index=False)
payments.to_csv("../data/cleaned/payments_cleaned.csv", index=False)
reviews.to_csv("../data/cleaned/reviews_cleaned.csv", index=False)
sellers.to_csv("../data/cleaned/sellers_cleaned.csv", index=False)
categories.to_csv("../data/cleaned/categories_cleaned.csv", index=False)
geolocation.to_csv("../data/cleaned/geolocation_cleaned.csv", index=False)
```
